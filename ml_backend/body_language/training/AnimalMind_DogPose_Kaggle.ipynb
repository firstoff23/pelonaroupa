# PeloNaRoupa — Dog Pose 24 Keypoints

Free Kaggle GPU training notebook. This stage trains the canine pose estimator; it does not yet train behavioral interpretation.

## Smoke test first
Run the 1-epoch cell before any long training. It verifies the current Ultralytics Dog-Pose download, 24-keypoint format, GPU, validation, and checkpoint creation.

```python
!pip -q install -U ultralytics
import torch
print('Torch:', torch.__version__)
print('CUDA:', torch.cuda.is_available())
if torch.cuda.is_available():
    print('GPU:', torch.cuda.get_device_name(0))
```

```python
from ultralytics import YOLO

model = YOLO('yolo26n-pose.pt')

results = model.train(
    data='https://raw.githubusercontent.com/ultralytics/ultralytics/main/ultralytics/cfg/datasets/dog-pose.yaml',
    epochs=1,
    imgsz=640,
    batch=8,
    device=0,
    workers=2,
    amp=True,
    cache=False,
    patience=15,
    project='/kaggle/working/animalmind',
    name='smoke-test',
    plots=True,
    save=True,
)

print('Smoke test concluído.')
print(results)
```
