# AnimalMind — Dog Pose 24 Keypoints

Free Kaggle GPU training notebook. This stage trains the canine pose estimator; it does not yet train behavioral interpretation.

```python
!pip -q install -U ultralytics
import torch
print('Torch:', torch.__version__)
print('CUDA:', torch.cuda.is_available())
if torch.cuda.is_available(): print('GPU:', torch.cuda.get_device_name(0))
```

```python
from ultralytics import YOLO
model = YOLO('yolo26n-pose.pt')
results = model.train(
    data='dog-pose.yaml',
    epochs=100,
    imgsz=640,
    batch=16,
    device=0,
    workers=2,
    patience=15,
    amp=True,
    cache=False,
    project='/kaggle/working/animalmind',
    name='dog-pose-24kp',
    plots=True,
    save=True,
)
print(results)
```
